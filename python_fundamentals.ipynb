## 3. Print, Input & String Formatting
- `print()` → shows output
- `input()` → asks the user for text — **always returns a `str`**, even for numbers
- **f-strings** → the clean way to combine variables into a sentence: `f"Hi {name}"`

```python

print(f"Hello, {name}!")
```

````python
# TODO: Build a mini "About Me" generator

name = input("What's your name? ")
age = input("How old are you? ")   # this comes back as a STRING
hobby = input("What's your favorite hobby? ")


# 1. Print an f-string introduction using name, age, and hobby

print(f"Hello I am {name}, I am {age} years old, and my favorite hobby is {hobby}.")

# 2. BONUS: try age + 5 directly (it breaks). Fix it with int(age) + 5
print(int(age) + 5)
````

## 4. Operators
- **Arithmetic:** `+ - * / // % **`
- **Comparison:** `== != > < >= <=` → always returns a `bool`
- **Logical:** `and`, `or`, `not` → combine conditions

````python
# TODO: Simple grade checker

score = 85
passing_score = 75

# 1. Check if the student passed (comparison operator)
passed = score > passing_score

# 2. Check if the score is even (modulo)
is_even = score == score % 2

# 3. Check if the student passed AND scored above 80 ('and')
honor_roll = score > passing_score and score > 80

print(passed, is_even, honor_roll)
````

## 5. Control Flow
- **Conditionals** `if / elif / else` → choose a path based on a condition
- **Loops** `for` (repeat a known number of times) / `while` (repeat until a condition changes)

```python
if score >= 75:
    print("Passed")
else:
    print("Failed")
```

````python
# TODO: Build a simple "bouncer" program

age = 18

# 1. if/elif/else:
#    "Welcome in!" if age >= 18
#    "Sorry, come back next year." if age is 17
#    "You're too young." otherwise

if (age >= 18):
    print("Welcome in!")
elif age == 17:
    print ("Sorry, come back next year.")
else:
    print("You're too young.")





# 2. BONUS: use a for loop to count down from 5 to 1, then print "Doors open!"
# countdown = 

for i in range (5, 0, -1):
    print(i)
print("Doors open!")
````
